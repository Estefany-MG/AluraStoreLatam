# 🏬 Desafío Alura Store: Análisis de Ventas
En este notebook analizaremos el rendimiento de las cuatro sucursales de Alura Store para ayudar al Señor Juan a decidir cuál de ellas vender.

## 📥 Importación de datos y preparación
Extraemos los datos desde el repositorio de GitHub y los agrupamos para facilitar el análisis.

```python
import pandas as pd
import matplotlib.pyplot as plt

# URLs proporcionadas en el desafío
url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

# Lectura de archivos CSV
tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

# Agrupamos en un diccionario para usar bucles for y no repetir código
tiendas = {
    'Tienda 1': tienda,
    'Tienda 2': tienda2,
    'Tienda 3': tienda3,
    'Tienda 4': tienda4
}

print("Datos cargados correctamente.")
tienda.head(3)
```

## 1. Análisis de facturación
Calculamos el ingreso total generado por cada tienda.

```python
facturacion_total = {}

print("💰 FACTURACIÓN TOTAL POR TIENDA 💰")
for nombre, df in tiendas.items():
    total = df['Precio'].sum()
    facturacion_total[nombre] = total
    print(f"{nombre}: ${total:,.2f}")

# Visualización
plt.figure(figsize=(8,5))
plt.bar(facturacion_total.keys(), facturacion_total.values(), color=['#4C72B0', '#55A868', '#C44E52', '#8172B2'])
plt.title('Facturación Total por Tienda')
plt.ylabel('Ingresos Totales ($)')
plt.show()
```

## 2. Ventas por categoría
Identificamos los productos que más se venden según su categoría.

```python
print("📦 TOP 3 CATEGORÍAS MÁS VENDIDAS POR TIENDA 📦")
for nombre, df in tiendas.items():
    print(f"\n--- {nombre} ---")
    top_categorias = df['Categoría del Producto'].value_counts().head(3)
    print(top_categorias.to_string())
```

## 3. Calificación promedio de la tienda
Evaluamos la satisfacción del cliente en cada sucursal.

```python
print("⭐ CALIFICACIÓN PROMEDIO POR TIENDA ⭐")
for nombre, df in tiendas.items():
    promedio = df['Calificación'].mean()
    print(f"{nombre}: {promedio:.2f} / 5.00")
```

## 4. Productos más y menos vendidos
Identificamos el producto estrella y el menos popular de cada inventario.

```python
print("🛒 PRODUCTOS MÁS Y MENOS VENDIDOS POR TIENDA 🛒")
for nombre, df in tiendas.items():
    ventas_producto = df['Producto'].value_counts()
    
    mas_vendido = ventas_producto.idxmax()
    cant_mas = ventas_producto.max()
    
    menos_vendido = ventas_producto.idxmin()
    cant_menos = ventas_producto.min()
    
    print(f"\n--- {nombre} ---")
    print(f"🔼 Más vendido: {mas_vendido} ({cant_mas} unidades)")
    print(f"🔽 Menos vendido: {menos_vendido} ({cant_menos} unidades)")
```

## 5. Envío promedio por tienda
Analizamos los costos logísticos promedio.

```python
print("🚚 COSTO DE ENVÍO PROMEDIO POR TIENDA 🚚")
for nombre, df in tiendas.items():
    envio_promedio = df['Costo de envío'].mean()
    print(f"{nombre}: ${envio_promedio:,.2f}")
```
